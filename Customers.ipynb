```python
import pandas as pd
import numpy as np
from pathlib import Path

PROJECT_DIR = Path.cwd().parent

file_path = (
    PROJECT_DIR
    / "data"
    / "Final_Version"
    / "Raw"
    / "Customers.csv"
)

print("File exists:", file_path.exists())

df = pd.read_csv(file_path)
pd.set_option('display.max.rows',None)
pd.set_option('display.max.columns',None)
df
df=df.sort_values('CustomerID')
df
print("Rows saved:", len(df))
```

```python
#To view the duplicates
duplicates = df[df.duplicated("CustomerID", keep=False)] \
            .sort_values("CustomerID")
duplicates
#To view the duplicates
```

```python

#last value value to be deleted
protected=[1095,
1103,
1134,
1125]
#first value to be deleted

to_drop=df.duplicated(subset=['CustomerID'],keep='first') & ~df['CustomerID'].isin(protected) #keep the first one for not in protected
#gives          the duplicated ids(second one)            &   not in protected list
#(second duplicate ids to be deleted and excludes the protected ids)(the ones to be deleted basically)

to_drop2=df.duplicated(subset=['CustomerID'],keep='last') & df['CustomerID'].isin(protected) #second one for protected
#gives          the duplicated ids(first one)            &  in protected list
#(first duplicate ids to be deleted and includes the protected ids)(the ones to be deleted basically)

all_to_drop=to_drop|to_drop2 #using or condition to combine all the ids to be deleted

df_clean=df[~all_to_drop] #creating a list which has non-deleted ids
df_dirty=df[all_to_drop]
"""just to review the IDs and check what must be deleted
to_view= df['CustomerID'].isin(protected) #view only the ids to be reviewed
df_view = df[to_view] 
df_view.sort_values('CustomerID')
"""

AUDIT_DIR = PROJECT_DIR / "data" / "Final_Version" / "Audit"
AUDIT_DIR.mkdir(parents=True, exist_ok=True)

audit_file = AUDIT_DIR / "Customers_Audit.csv"

df_dirty.to_csv(audit_file, index=False)


print("Saved successfully:", audit_file)
print("Rows saved:", len(df_dirty))
CLEAN_DIR = PROJECT_DIR / "data" / "Final_Version" / "Clean"
CLEAN_DIR.mkdir(parents=True, exist_ok=True)

clean_file = CLEAN_DIR / "Customers.csv"

df_clean.to_csv(clean_file, index=False)

df_clean.sort_values('CustomerID')
df=df_clean.sort_values('CustomerID')
df
print("Rows saved:", len(df))

#All the duplicate customer ids are deleted.



```

```python
df['IncorporationDate']=pd.to_datetime(df['IncorporationDate'],
format="mixed",
dayfirst=True,
errors="coerce")
df
                                     
```
