```python
import pandas as pd
import numpy as np
from pathlib import Path

PROJECT_DIR = Path.cwd().parent

file_path = (
    PROJECT_DIR
    / "data"
    / "Final_Version"
    / "Raw"
    / "Approvals.csv"
)

print("File exists:", file_path.exists())

df = pd.read_csv(file_path)

pd.set_option("display.max.rows",None)
pd.set_option("display.max.columns",None)

df.sort_values("ApprovalID")
print("Rows saved:", len(df))
```

```python
df["ApprovalDate"]=pd.to_datetime(df["ApprovalDate"],
                                  format="mixed",
                                  dayfirst=True,
                                  errors="coerce")
df
```

```python
df["ApprovalStatus"]=(df["ApprovalStatus"]
                      .str.strip()
                      .str.title())

df["ApprovedBy"]=(df["ApprovedBy"]
                      .str.strip()
                      .str.title())
df

```

```python
CLEAN_DIR = PROJECT_DIR / "data" / "Final_Version" / "Clean"
output_file = CLEAN_DIR / "Facilities_Clean.csv"
df1=pd.read_csv(output_file)
df1
```

```python
df[~df['FacilityID'].isin(df1['FacilityID'])]

df['Facilities_Exists']=df['FacilityID'].isin(df1['FacilityID'])
df
```
