```python
df["FacilityType"]=(df["FacilityType"]
                    .str.strip()
                    .str.title()
)
df
```

```python
df["AmountApproved"]=(df["AmountApproved"].abs())
df
```

```python
df["StartDate"]=pd.to_datetime(df["StartDate"],
                              format="mixed",
                               dayfirst=True,
                               errors="coerce")
df
```

```python
df["TenorMonths"]=pd.to_numeric(df["TenorMonths"],errors="coerce")
df
```

```python
df=df.replace(r'^\s*$',np.nan,regex=True)
df
```
