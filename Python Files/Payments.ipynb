```python
import pandas as pd
import numpy as np
from pathlib import Path

PROJECT_DIR = Path.cwd().parent

file_path = (
    PROJECT_DIR
    / "data"
    / "Final_Version"
    / "Raw"
    / "Payments.csv"
)
print("File exists:", file_path.exists())
df = pd.read_csv(file_path)
print("Rows saved:", len(df))

CLEAN_DIR = PROJECT_DIR / "data" / "Final_Version" / "Clean"
output_file = CLEAN_DIR / "Facilities_Clean.csv"

print("File exists:", output_file.exists())

df2 = pd.read_csv(output_file)
print("Rows saved:", len(df2))

pd.set_option("display.max.rows",None)
pd.set_option("display.max.columns",None)

df.sort_values("PaymentID")

```

```python
df["PaymentID"].is_unique
df["FacilityID"].isin(df2["FacilityID"])
```

```python
df["Status"]=(df["Status"]
                      .str.strip()
                      .str.title())
df
```

```python
df["PaymentID"].is_unique
```

```python
df['PaymentDate']=pd.to_datetime(df['PaymentDate'],
format="mixed",
dayfirst=True,
errors="coerce")
df
```

```python
df["FacilityIDExists"]=df["FacilityID"].isin(df2["FacilityID"])
df
```

```python
df["NegativeValueExists"]=df["Amount"]<0
df
```

```python
df["PaymentDateMissingOrInvalid"] = df["PaymentDate"].isna()
df["AmountMissing"] = df["Amount"].isna()
df["StatusMissing"] = df["Status"].isna()
```

```python
PROJECT_DIR = Path.cwd().parent

CLEAN_DIR = PROJECT_DIR / "data" / "Final_Version" / "Clean"
CLEAN_DIR.mkdir(parents=True, exist_ok=True)

output_file = CLEAN_DIR / "Payments_Clean.csv"

df.to_csv(output_file, index=False)

print("Saved successfully:", output_file)
print("Rows saved:", len(df))
```
