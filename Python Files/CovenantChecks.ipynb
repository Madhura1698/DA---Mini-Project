```python
import pandas as pd
import numpy as np
from pathlib import Path

PROJECT_DIR = Path.cwd().parent

file_path = (
    PROJECT_DIR
    / "data"
    / "Final_Version"
    / "Raw"
    / "CovenantChecks.csv"
)

print("File exists:", file_path.exists())
df = pd.read_csv(file_path)

CLEAN_DIR = PROJECT_DIR / "data" / "Final_Version" / "Clean"
output_file = CLEAN_DIR / "Facilities_Clean.csv"

print("File exists:", output_file.exists())
df2 = pd.read_csv(output_file)

pd.set_option("display.max.rows",None)
pd.set_option("display.max.columns",None)

df.sort_values("CovenantID")
print("Rows saved:", len(df))
```

```python
df3=df.duplicated(subset="CovenantID",keep=False)
df3
df["DuplicateExistsFlag"]=df.duplicated(subset="CovenantID",keep=False)
df
```

```python
df["FacilityExistsFlag"]=df["FacilityID"].isin(df2["FacilityID"])
df
```

```python
df['LastCheckedDate']=pd.to_datetime(df['LastCheckedDate'],
format="mixed",
dayfirst=True,
errors="coerce")
df
```

```python
df["Threshold"]=pd.to_numeric(df["Threshold"],errors="coerce")
df
```

```python
df["CovenantType"]=(df["CovenantType"]
                      .str.strip()
                      .str.upper())
df
```

```python
PROJECT_DIR = Path.cwd().parent

CLEAN_DIR = PROJECT_DIR / "data" / "Final_Version" / "Clean"
CLEAN_DIR.mkdir(parents=True, exist_ok=True)

output_file = CLEAN_DIR / "CovenantChecks_Clean.csv"

df.to_csv(output_file, index=False)

print("Saved successfully:", output_file)
print("Rows saved:", len(df))
```
